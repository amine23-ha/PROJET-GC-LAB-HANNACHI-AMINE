## Le but

Créer une petite carte avec des zones. Le programme doit dire si un point est dans une zone, calculer sa surface, puis vérifier les résultats sur un dessin. Je vais aussi comparer une aire exacte à une estimation avec des points au hasard.

Ce projet reprend surtout les fonctions `aire`, `dedans` et `tracer_polygone` du TP6.

## 1. Les fonctions du TP6

La fonction `aire` utilise la formule du lacet. La fonction `dedans` compte les croisements avec un rayon horizontal. Un nombre impair de croisements veut dire que le point est dedans.

```python
import random
import matplotlib.pyplot as plt

def aire(polygone):
    somme = 0
    for i in range(len(polygone)):
        x1, y1 = polygone[i]
        x2, y2 = polygone[(i + 1) % len(polygone)]
        somme = somme + x1 * y2 - x2 * y1
    return abs(somme) / 2

def dedans(P, polygone):
    x, y = P
    resultat = False
    for i in range(len(polygone)):
        x1, y1 = polygone[i]
        x2, y2 = polygone[(i + 1) % len(polygone)]
        if (y1 > y) != (y2 > y):
            intersection = x1 + (y - y1) * (x2 - x1) / (y2 - y1)
            if x < intersection:
                resultat = not resultat
    return resultat

def tracer_polygone(polygone, points):
    xs = [p[0] for p in polygone] + [polygone[0][0]]
    ys = [p[1] for p in polygone] + [polygone[0][1]]
    plt.figure(figsize=(5, 5))
    plt.plot(xs, ys, color="black")
    for P in points:
        couleur = "green" if dedans(P, polygone) else "red"
        plt.scatter(P[0], P[1], color=couleur)
    plt.axis("equal")
    plt.title("Vert : dedans, rouge : dehors")
    plt.show()
```

## 2. Tester une zone en forme de L

Le L est un polygone non convexe. Sa surface doit valoir 7 : deux rectangles de surface 4, avec un carré commun de surface 1. Le point (2, 2) est dans le creux, donc il doit être dehors.

```python
L = [(0, 0), (4, 0), (4, 1), (1, 1), (1, 4), (0, 4)]
points = [(0.5, 0.5), (3, 0.5), (2, 2)]

print("Aire du L :", aire(L))
for P in points:
    print(P, dedans(P, L))
tracer_polygone(L, points)
```

## 3. Chercher la zone qui contient un point

On ajoute un carré à côté du L. La fonction renvoie le numéro de la zone ou `None` si le point ne se trouve dans aucune.

```python
def zone_de(P, zones):
    for i in range(len(zones)):
        if dedans(P, zones[i]):
            return i
    return None

carre = [(5, 0), (7, 0), (7, 2), (5, 2)]
zones = [L, carre]

print(zone_de((0.5, 0.5), zones))
print(zone_de((6, 1), zones))
print(zone_de((2, 2), zones))
```

## 4. Estimer la surface avec des points au hasard

Le rectangle qui entoure le L mesure 4 × 4, donc sa surface est 16. Je tire des points au hasard dans ce rectangle et je compte ceux qui sont dans le L. Je compare ensuite l’estimation avec l’aire exacte 7. Cette idée vient de la méthode de Monte-Carlo du TP6.

```python
def estimation_aire(nombre_points):
    compteur = 0
    for i in range(nombre_points):
        x = random.uniform(0, 4)
        y = random.uniform(0, 4)
        if dedans((x, y), L):
            compteur = compteur + 1
    return 16 * compteur / nombre_points

random.seed(1)
for n in [100, 1000, 10000]:
    estimation = estimation_aire(n)
    print(n, "points :", round(estimation, 3), "erreur :", round(abs(7 - estimation), 3))
```

## 5. Vérifier plusieurs résultats

Un seul exemple ne suffit pas. Je vérifie aussi un carré, un point dans le creux et un point hors des deux zones.

```python
print("Aire du L correcte :", aire(L) == 7)
print("Aire du carre correcte :", aire(carre) == 4)
print("Point dans le L :", dedans((0.5, 0.5), L))
print("Point dans le creux :", dedans((2, 2), L))
print("Zone du carre :", zone_de((6, 1), zones))
print("Aucune zone :", zone_de((4.5, 1), zones))
```

## Conclusion

La formule du lacet donne une aire de 7 pour le L. Le dessin montre que les deux points verts sont dedans, mais que le point rouge placé dans le creux est dehors. La fonction `zone_de` retrouve aussi la bonne zone.

Avec les points au hasard, on obtient une estimation de l’aire, pas toujours 7 exactement. En général, avec davantage de points, le résultat se rapproche de 7, mais ce n’est pas garanti à chaque essai.

**Limites :** le programme utilise une carte très simple. Les points exactement sur un côté ou sur un sommet demanderaient une vérification supplémentaire. La méthode aléatoire est aussi moins précise que la formule du lacet pour un polygone connu.
